### We'll define some helper functions here: 
(1) to calculate MSE, and 
(2) to display the improvement of a GBM on test data as additional trees are constructed.

```python
def mse(y, y_hat):
    """
    Calculate the MSE of a model given actual and predicted values
    """
    # calculate the residual error for each individual record
    resid = y - y_hat
    # square the residual (hence "squared error")
    sq_resid = resid**2
    # calculate the sum of squared errors
    SSR = sum(sq_resid)
    # divide by the number of records to get the mean squared error
    MSE = SSR / y.shape[0]
    return MSE

def gbm_progress(model_name, X_test, y_test):
    """
    Plot the 'progress' of a GBM on a test set to assess robustness
    """
    test_error = np.zeros_like(model_name.train_score_)
    for idx, y_ in enumerate(model_name.staged_predict(X_test)):
        test_error[idx] = np.mean((y_test - y_)**2)

    plot_idx = np.arange(model_name.train_score_.shape[0])
    ax = subplots(figsize=(8,8))[1]
    ax.plot(plot_idx,
            model_name.train_score_,
            'b',
            label='Training')
    ax.plot(plot_idx,
            test_error,
            'r',
            label='Test')
    ax.legend();
```

### Build a basic GBM first.

```python
# Documentation: https://scikit-learn.org/dev/modules/generated/sklearn.ensemble.GradientBoostingRegressor.html

gbm_basic = GBR(learning_rate=0.1,
               n_estimators=250,
               max_depth=10,
               min_samples_split=2,
               subsample=1,
               random_state=seed)

gbm_basic.fit(X_train[indep_vars], y_train)
```

```python
y_hat_basic = gbm_basic.predict(X_test[indep_vars])

mse_basic = mse(y_test, y_hat_basic)
print('test mse: ',mse_basic)
```

### When we plot the performance of the model on training and test data, we see that performance no longer improves after a certain number of trees (in this case, between 50-100 trees).

```python
gbm_progress(gbm_basic, X_test[indep_vars], y_test)
```

### Now you'll have the chance to choose optimal hyperparameters to build a better GBM.
Once the candidate hyperparameters are chosen, use 5-fold cross-validation to decide which collection of hyperparameters are best.

```python
# Build a stronger feature set for the final model.
def add_gbm_features(df, bedrooms_fill=None):
    out = df.copy()
    if bedrooms_fill is None:
        bedrooms_fill = out["total_bedrooms"].median()
    out["total_bedrooms"] = out["total_bedrooms"].fillna(bedrooms_fill)
    out["rooms_per_household"] = out["total_rooms"] / out["households"]
    out["population_per_household"] = out["population"] / out["households"]
    out["bedrooms_per_room"] = out["total_bedrooms"] / out["total_rooms"]
    for col in ["rooms_per_household", "population_per_household", "bedrooms_per_room"]:
        out[col] = out[col].replace([np.inf, -np.inf], np.nan)
        out[col] = out[col].fillna(out[col].median())
    return out, bedrooms_fill

gbm_train, bedrooms_fill = add_gbm_features(Train)
gbm_test, _ = add_gbm_features(Test, bedrooms_fill=bedrooms_fill)

indep_vars = ['longitude', 'latitude', 'housing_median_age', 'total_rooms',
              'total_bedrooms', 'population', 'households', 'median_income',
              'ocean_prox_inland', 'ocean_prox_nearocean', 'ocean_prox_nearbay',
              'rooms_per_household', 'population_per_household', 'bedrooms_per_room']

X_train = gbm_train[indep_vars]
X_test = gbm_test[indep_vars]
y_train = gbm_train['median_house_value']
y_test = gbm_test['median_house_value']

def get_best_gbm(X, y, results_path="gridsearch_results.pkl", force_retrain=False):
    """Load cached HGBR search results when available; otherwise run a compact 5-fold search."""

    search_version = "hgbm_cv_v1"

    if os.path.exists(results_path) and not force_retrain:
        try:
            data = joblib.load(results_path)
            model = data.get("best_model") if isinstance(data, dict) else None
            params = data.get("best_params") if isinstance(data, dict) else None
            if data.get("search_version") == search_version and model is not None and params is not None:
                print("Loaded cached GridSearchCV results.")
                return model, params
            print("Cached results do not match the current HGBR search. Running GridSearchCV...")
        except Exception as exc:
            print(f"Could not load cached results ({exc}). Running GridSearchCV...")

    print("Running GridSearchCV...")

    gbm_cv = HGBR(random_state=seed, early_stopping=False)

    param_grid = {
        "learning_rate": [0.05],
        "max_depth": [None, 8],
        "max_leaf_nodes": [31],
        "min_samples_leaf": [10, 20],
        "max_iter": [500, 600]
    }

    grid_search = GridSearchCV(
        estimator=gbm_cv,
        param_grid=param_grid,
        cv=5,
        scoring="neg_mean_squared_error",
        n_jobs=-1,
        verbose=0
    )

    grid_search.fit(X, y)

    best_model = grid_search.best_estimator_
    best_params = grid_search.best_params_

    joblib.dump({
        "best_model": best_model,
        "best_params": best_params,
        "cv_results": grid_search.cv_results_,
        "search_version": search_version
    }, results_path)

    print(f"Saved results to {results_path}")

    return best_model, best_params
```

```python
# Identify best model
best_model, best_params = get_best_gbm(X_train[indep_vars], y_train, force_retrain=False)

print("Best parameters:")
print(best_params)
```
